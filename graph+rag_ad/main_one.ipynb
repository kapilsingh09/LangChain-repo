```python
from pydantic import BaseModel, Field
from typing import List


class Entity(BaseModel):
    name: str
    type: str
    description: str


class Relationship(BaseModel):
    source: str
    relationship: str
    target: str


class GraphExtraction(BaseModel):
    entities: List[Entity]
    relationships: List[Relationship]
```
